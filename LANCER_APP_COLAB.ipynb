## 1. Importer le ZIP de l'application

````python
from google.colab import files
import zipfile, shutil, sys
from pathlib import Path

print("Importez SmartTrip_App_Complet.zip")
uploaded = files.upload()
zip_name = next(name for name in uploaded if name.lower().endswith(".zip"))

root = Path("/content/smarttrip")
if root.exists():
    shutil.rmtree(root)
root.mkdir()

with zipfile.ZipFile(zip_name, "r") as z:
    z.extractall(root)

matches = list(root.rglob("SmartTrip_LLM_Agent_App"))
PROJECT_DIR = matches[0] if matches else next(p for p in root.iterdir() if p.is_dir())
sys.path.insert(0, str(PROJECT_DIR))
sys.path.insert(0, str(PROJECT_DIR / "app"))

print("Projet :", PROJECT_DIR)
````

## 2. Installer les dépendances de l'application

````python
!pip -q install -r "$PROJECT_DIR/requirements_app.txt"
````

## 3. Lancer l'application

Colab affichera un **lien public Gradio**. Clique dessus pour ouvrir l'application.


````python
import os
os.chdir(PROJECT_DIR)

from app.gradio_app import demo
demo.launch(share=True, debug=False)
````
