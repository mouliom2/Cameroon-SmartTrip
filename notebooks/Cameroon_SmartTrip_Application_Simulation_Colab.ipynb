# 🇨🇲 Cameroon SmartTrip — Application de simulation dans Colab

Ce notebook lance une **vraie interface web** dans Google Colab.

L'application comporte trois onglets :

1. **Planifier un séjour**
2. **Simulation de plusieurs profils**
3. **Stress test du budget**


## 1. Importer le ZIP complet

```python
from google.colab import files
import zipfile, shutil, sys
from pathlib import Path

print("Importez : Projet_03_Cameroon_SmartTrip_APPLICATION_SIMULATION_COLAB.zip")
uploaded = files.upload()

zip_name = next(name for name in uploaded if name.lower().endswith(".zip"))

extract_dir = Path("/content/cameroon_smarttrip")
if extract_dir.exists():
    shutil.rmtree(extract_dir)
extract_dir.mkdir()

with zipfile.ZipFile(zip_name, "r") as z:
    z.extractall(extract_dir)

matches = list(extract_dir.rglob("Cameroon-SmartTrip-Colab"))
PROJECT_DIR = matches[0] if matches else next(p for p in extract_dir.iterdir() if p.is_dir())
APP_DIR = PROJECT_DIR / "app"

sys.path.insert(0, str(APP_DIR))

print("Projet :", PROJECT_DIR)
```

## 2. Installer les dépendances

```python
import subprocess, sys

subprocess.check_call([
    sys.executable, "-m", "pip", "install", "-q",
    "-r", str(PROJECT_DIR / "requirements_colab.txt")
])

print("Dépendances installées.")
```

## 3. Lancer l'application

Après exécution, Colab affichera un **lien public Gradio**.  
Clique dessus pour ouvrir l'application complète dans ton navigateur.


```python
import os
os.chdir(PROJECT_DIR)

from gradio_app import demo

demo.launch(share=True, debug=False)
```

## Utilisation

Dans l'application :

- **Planifier un séjour** : tu testes une demande individuelle.
- **Simulation de plusieurs profils** : tu génères 5 à 100 voyageurs simulés.
- **Stress test budget** : tu gardes le même profil et tu fais varier le budget.

Tu peux ainsi **simuler le comportement de SmartTrip depuis une vraie interface**, sans modifier le code.
